```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
def responder_a_los_polillas(numero):
    
    resultado = []
    for ii in range(numero):
        if ii%2 == 0:
            resultado_temporal = ii/100
#             print(ii, resultado_temporal)
            resultado.append(resultado_temporal)
        else:
            resultado_temporal = np.cos(0.1*ii)
#             print(ii, resultado_temporal)
            resultado.append(resultado_temporal)
    
    return resultado
```

```python
resultado = responder_a_los_polillas(100)
```

```python
fig, ax = plt.subplots(1,1)

ax.scatter(np.arange(100), resultado, s=0.5)
ax.plot(np.arange(100), resultado, lw=0.5)
```
